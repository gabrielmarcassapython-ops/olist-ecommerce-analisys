```python
# Importando as bibliotecas
import pandas as pd          # manipulação de dados (a mais importante)
import matplotlib.pyplot as plt  # criação de gráficos
import seaborn as sns        # gráficos mais bonitos e fáceis
import plotly.express as px  # gráficos interativos

# Configurações visuais
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')

print("Bibliotecas carregadas com sucesso! ✅")
```

```python
# Definindo o caminho para os dados
caminho = r"C:/Users/Acer/primeiro-projeto-kaggle-github/data/raw/"

# Carregando os arquivos CSV
orders = pd.read_csv(caminho + "olist_orders_dataset.csv")
items = pd.read_csv(caminho + "olist_order_items_dataset.csv")
customers = pd.read_csv(caminho + "olist_customers_dataset.csv")
products = pd.read_csv(caminho + "olist_products_dataset.csv")
reviews = pd.read_csv(caminho + "olist_order_reviews_dataset.csv")
sellers = pd.read_csv(caminho + "olist_sellers_dataset.csv")
payments = pd.read_csv(caminho + "olist_order_payments_dataset.csv")

print("Dados carregados com sucesso! ✅")
```

```python
# Renomeando as colunas para português

orders = orders.rename(columns={
    'order_id': 'pedido_id',
    'customer_id': 'cliente_id',
    'order_status': 'status',
    'order_purchase_timestamp': 'data_compra',
    'order_approved_at': 'data_aprovacao',
    'order_delivered_carrier_date': 'data_envio',
    'order_delivered_customer_date': 'data_entrega',
    'order_estimated_delivery_date': 'data_estimada_entrega'
})

customers = customers.rename(columns={
    'customer_id': 'cliente_id',
    'customer_unique_id': 'cliente_unico_id',
    'customer_zip_code_prefix': 'cep',
    'customer_city': 'cidade',
    'customer_state': 'estado'
})

items = items.rename(columns={
    'order_id': 'pedido_id',
    'order_item_id': 'item_id',
    'product_id': 'produto_id',
    'seller_id': 'vendedor_id',
    'shipping_limit_date': 'data_limite_envio',
    'price': 'preco',
    'freight_value': 'frete'
})

products = products.rename(columns={
    'product_id': 'produto_id',
    'product_category_name': 'categoria',
    'product_name_lenght': 'tamanho_nome',
    'product_description_lenght': 'tamanho_descricao',
    'product_photos_qty': 'qtd_fotos',
    'product_weight_g': 'peso_g',
    'product_length_cm': 'comprimento_cm',
    'product_height_cm': 'altura_cm',
    'product_width_cm': 'largura_cm'
})

reviews = reviews.rename(columns={
    'review_id': 'avaliacao_id',
    'order_id': 'pedido_id',
    'review_score': 'nota',
    'review_comment_title': 'titulo_comentario',
    'review_comment_message': 'comentario',
    'review_creation_date': 'data_avaliacao',
    'review_answer_timestamp': 'data_resposta'
})

sellers = sellers.rename(columns={
    'seller_id': 'vendedor_id',
    'seller_zip_code_prefix': 'cep',
    'seller_city': 'cidade',
    'seller_state': 'estado'
})

payments = payments.rename(columns={
    'order_id': 'pedido_id',
    'payment_sequential': 'sequencia',
    'payment_type': 'tipo_pagamento',
    'payment_installments': 'parcelas',
    'payment_value': 'valor'
})

print("Colunas renomeadas com sucesso! ✅")
```

```python
# Contando clientes por estado
clientes_por_estado = customers.groupby('estado')['cliente_id'].count().sort_values(ascending=False)

# Criando o gráfico
plt.figure(figsize=(14, 6))
clientes_por_estado.plot(kind='bar', color='steelblue')
plt.title('Número de Clientes por Estado', fontsize=16)
plt.xlabel('Estado')
plt.ylabel('Número de Clientes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Juntando a tabela de itens com a de produtos
itens_produtos = items.merge(products[['produto_id', 'categoria']], on='produto_id')

# Contando vendas por categoria e pegando as top 15
top_categorias = itens_produtos.groupby('categoria')['pedido_id'].count().sort_values(ascending=False).head(15)

# Criando o gráfico
plt.figure(figsize=(14, 6))
top_categorias.plot(kind='bar', color='steelblue')
plt.title('Top 15 Categorias Mais Vendidas', fontsize=16)
plt.xlabel('Categoria')
plt.ylabel('Número de Vendas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Contando quantas avaliações tem cada nota
notas = reviews.groupby('nota')['avaliacao_id'].count()

# Calculando a média geral
media = reviews['nota'].mean()

# Criando o gráfico
plt.figure(figsize=(10, 6))
notas.plot(kind='bar', color=['#e74c3c', '#e67e22', '#f1c40f', '#2ecc71', '#27ae60'])
plt.title(f'Distribuição das Avaliações dos Clientes\nMédia Geral: {media:.2f}', fontsize=16)
plt.xlabel('Nota')
plt.ylabel('Número de Avaliações')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Convertendo a coluna de data para o formato de data do Python
orders['data_compra'] = pd.to_datetime(orders['data_compra'])

# Extraindo o mês e ano de cada pedido
orders['mes_ano'] = orders['data_compra'].dt.to_period('M')

# Contando pedidos por mês
pedidos_por_mes = orders.groupby('mes_ano')['pedido_id'].count()

# Criando o gráfico
plt.figure(figsize=(14, 6))
pedidos_por_mes.plot(kind='line', color='steelblue', linewidth=2, marker='o')
plt.title('Volume de Pedidos por Mês', fontsize=16)
plt.xlabel('Mês/Ano')
plt.ylabel('Número de Pedidos')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Convertendo as datas para o formato correto
orders['data_entrega'] = pd.to_datetime(orders['data_entrega'])
orders['data_compra'] = pd.to_datetime(orders['data_compra'])

# Calculando o tempo de entrega em dias
orders['dias_entrega'] = (orders['data_entrega'] - orders['data_compra']).dt.days

# Juntando com a tabela de clientes para ter o estado
orders_clientes = orders.merge(customers[['cliente_id', 'estado']], on='cliente_id')

# Calculando a média de dias por estado e ordenando
entrega_por_estado = orders_clientes.groupby('estado')['dias_entrega'].mean().sort_values(ascending=True).round(1)

# Criando o gráfico
plt.figure(figsize=(14, 6))
entrega_por_estado.plot(kind='bar', color='steelblue')
plt.title('Tempo Médio de Entrega por Estado (em dias)', fontsize=16)
plt.xlabel('Estado')
plt.ylabel('Dias')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
